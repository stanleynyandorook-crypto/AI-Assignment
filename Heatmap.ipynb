1.import required libraries

```python
import pandas as pd
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.metrics import(
    mean_squared_error,
    r2_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
```

2.load the daibetes dataset

```python
diabetes=load_diabetes(as_frame=True)
df=diabetes.frame
df.head()
```

Draw correlation haetmap

```python
plt.figure(figsize=(12,10))
sns.heatmap(df.corr(),annot=True)
plt.title("Correlation Heatmap")
plt.show()
```


1.   1.which feature has the strongest positive correlation with the target?
The feature with the strongest positive correlation with the target is 'bmi' with a correlation of 0.586
2.  .Which feature has the weakest corretion?
The feature with the weakest correlation (closest to zero) with the target is 'sex' with a correlation of 0.043.



Separate the independent variables (X) from he target variable(y)

```python
x=df.drop('target',axis=1)
y=df['target']
```

Split the database into training and testing data

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

Train the linear regression model

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

Make predictons

```python
predictions = model.predict(x_test)
print(predictions)
```

Evaluate the model

```python
mse = mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)
print(f"Mean Squared Error: {mse}")
print(f"R-squared: {r2}")
```

Draw a scatter plot

```python
plt.figure(figsize=(7,5))
plt.scatter(y_test,predictions)
plt.xlabel("Actual Target")
plt.ylabel("Predicted Target")
plt.title("Actual vs Predicted Target")
plt.show()
```
